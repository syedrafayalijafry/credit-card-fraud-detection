# EDA Summary

## Key Findings

### Dataset Overview
- The IEEE-CIS Fraud Detection dataset contains **590,540 transactions** and over **430 features** after merging the transaction and identity datasets.
- The dataset consists of numerical, categorical, and engineered features grouped into Transaction, Card, Address, Distance, C, D, M, V, and Identity feature groups.

### Target Variable
- The target variable (`isFraud`) is **highly imbalanced**, with legitimate transactions significantly outnumbering fraudulent transactions.
- Accuracy alone is not an appropriate evaluation metric; therefore, Precision, Recall, F1-Score, and ROC-AUC will be used during model evaluation.

### Missing Values
- Missing values are widespread across the dataset.
- Several feature groups, particularly the **V** and **Identity (`id_*`)** features, contain a high percentage of missing values.
- Features will not be removed solely based on missing value percentage, as some highly informative features also contain substantial missing data.

### Numerical Features
- **TransactionAmt** exhibits a highly right-skewed distribution with numerous extreme outliers.
- **TransactionDT** represents elapsed time rather than an actual timestamp and may provide useful temporal information.
- Distance-related features (`dist1` and `dist2`) contain missing values and require preprocessing.

### Categorical Features
- Product categories, card information, email domains, and device-related features show varying distributions across the dataset.
- These categorical variables are expected to contribute meaningful information after appropriate encoding.

### Feature Groups
- **C Features:** No missing values but relatively weak individual correlations with the target.
- **D Features:** Contain varying levels of missing values and may capture temporal behavior.
- **M Features:** Binary categorical variables requiring encoding.
- **V Features:** Largest feature group with substantial missing values, yet several features demonstrate strong correlations with fraud.
- **Identity Features (`id_*`):** Include both numerical and categorical variables with considerable missing values that may provide valuable predictive information.

### Correlation Analysis
- Most individual features exhibit weak linear correlations with the target.
- Several engineered **V features** show comparatively stronger positive correlations with fraudulent transactions.
- Low correlation does not necessarily imply low predictive power, particularly for tree-based machine learning models.
