```python
plt.figure(figsize=(6, 4))

sns.countplot(
    data=df,
    x="isFraud"
)

plt.title("Distribution of Fraudulent and Legitimate Transactions")
plt.xlabel("isFraud")
plt.ylabel("Count")

plt.show()
```

```python
fraud_percentage.plot(
    kind="bar",
    figsize=(6,4)
)

plt.title("Fraud Percentage")
plt.xlabel("Class")
plt.ylabel("Percentage (%)")

plt.xticks(rotation=0)

plt.show()
```

### Observations

- The dataset is highly imbalanced.
- Legitimate transactions make up the majority of the data.
- Fraudulent transactions represent only a small percentage.
- Accuracy alone is not an appropriate evaluation metric.
- During model evaluation, we will focus on:
  - Precision
  - Recall
  - F1-Score
  - ROC-AUC

```python
missing_values = df.isnull().sum()

missing_values
```

```python
missing_percentage = (
    df.isnull()
      .mean()
      .mul(100)
      .round(2)
)

missing_percentage
```

```python
missing_summary = pd.DataFrame({
    "Missing Values": missing_values,
    "Missing Percentage": missing_percentage
})

missing_summary = missing_summary[
    missing_summary["Missing Values"] > 0
].sort_values(
    by="Missing Percentage",
    ascending=False
)

missing_summary
```

```python
print(
    f"Columns with missing values: {missing_summary.shape[0]}"
)
```

```python
missing_summary.head(20)
```

```python
plt.figure(figsize=(12, 8))

missing_summary.head(20)["Missing Percentage"].plot(
    kind="bar"
)

plt.title("Top 20 Features with Highest Missing Percentage")
plt.xlabel("Features")
plt.ylabel("Missing Percentage (%)")

plt.xticks(rotation=90)

plt.show()
```

### Observations

- The IEEE-CIS dataset contains a significant number of missing values.
- Some features have a very high percentage of missing data.
- Missing values are not uniformly distributed across the dataset.
- No preprocessing decisions will be made during EDA.
- Missing value handling will be implemented later in `src/preprocessing.py`.
