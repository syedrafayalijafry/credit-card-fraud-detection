## Logistic Regression

```python
lr_model = train_logistic_regression(X_train, y_train)
```

```python
lr_result = evaluate_model(lr_model, X_test, y_test, "Logistic Regression")

lr_result
```

## Decision Tree

```python
dt_model = train_decision_tree(X_train, y_train)
```

```python
dt_result = evaluate_model(dt_model, X_test, y_test, "Decision Tree")

dt_result
```

## Random Forest

```python
rf_model = train_random_forest(X_train, y_train)
```

```python
rf_result = evaluate_model(rf_model, X_test, y_test, "Random Forest")
```

## XGBoost

```python
xgb_model = train_xgboost(X_train, y_train)
```

```python
xgb_result = evaluate_model(xgb_model, X_test, y_test, "XGBoost")
```

## LightGBM

```python
lgb_model = train_lightgbm(X_train, y_train)
```

```python
lgb_result = evaluate_model(lgb_model, X_test, y_test, "LightGBM")
```

## CatBoost

```python
cat_model = train_catboost(X_train, y_train)
```

```python
cat_result = evaluate_model(cat_model, X_test, y_test, "CatBoost")
```

## Compare All Models

```python
results = [
    lr_result,
    dt_result,
    rf_result,
    xgb_result,
    lgb_result,
    cat_result
]

results_df = compare_models(results)

results_df
```

## Best Model

## Model Training Summary

- Six classification models were trained and evaluated.
- ROC-AUC was used as the primary evaluation metric because the dataset is highly imbalanced.
- Precision, Recall, and F1-score were also considered.
- LightGBM achieved the highest ROC-AUC and provided the best balance between performance and training time.
- Therefore, LightGBM was selected for hyperparameter tuning in the next stage.
